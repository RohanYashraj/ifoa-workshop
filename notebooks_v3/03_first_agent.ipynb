## 2. Hello, agent — one tool, live trace

```python
from agno.agent import Agent
from agno.models.google import Gemini

def claim_frequency(region: str) -> float:
    """ABC Motor claim frequency for a region tier: Tier1, Tier2 or Tier3."""
    return {"Tier1": 0.069, "Tier2": 0.082, "Tier3": 0.091}[region]

agent = Agent(
    model=Gemini(id=MODEL),
    tools=[claim_frequency],
    show_tool_calls=True,
)
agent.print_response("Compare claim frequency for Tier-1 vs Tier-2. Which is higher and by how much?")
```

## 3. Read the trace
The model called the tool **twice** — nobody told it to. That's the ReAct loop deciding. The 19% gap is computed from tool results, not hallucinated.
